```python
import pandas as pd
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))
from python.data_cleaning import  (standardize_columns, clean_text_col,
                                  convert_datetime, convert_numeric,
                                  remove_duplicate, remove_missing,
                                  valid, saved_file, remove_col)
```

```python
df_cus = pd.read_csv("../data/raw/customer_master.csv")
df_cus
```

```python
df_cus_clean = standardize_columns(df_cus)
columns = ["customer_name", "gender","customer_segment","customer_city","customer_state","customer_country",	"region"]
df_cus_clean = clean_text_col(df_cus, columns)
df_cus_clean
```

# Cleaning sales_table

```python
df_sale = pd.read_csv("../data/raw/ecommerce_sales_customer_analytics_150k.csv")
df_sale
```

```python
df_sale[df_sale['customer_rating'].notnull()]['customer_rating']
```

```python
column = ['customer_name', 'customer_age', 'gender', 'customer_segment', 'customer_city', 'customer_state', 'customer_country', 'region', 'customer_postal_code', 'customer_type',
          'warehouse']

df_sale_clean = remove_col(df_sale, column)
df_sale_clean.info()
```

```python
columns = ['order_date', 'order_time']
txt_columns = df_sale_clean.select_dtypes(include=['string']).columns.tolist()
rules = {'delivery_days':'float64', 'estimated_delivery_days':'float64', 'customer_rating':'float64',
         'loyalty_points_earned':'int64', 'loyalty_points_redeemed':'int64', 'quantity':'int64', 'gross_sales' :'float64',
   'discount_amount':'float64',
   'tax_amount'     :'float64',
   'shipping_cost'  : 'float64',
   'net_sales'     :'float64',
    'product_cost'    :'float64',
    'profit'        :'float64',
    'profit_margin_percentage'  :'float64',
   'customer_lifetime_value'   :'float64',
   'is_repeat_customer' :'bool',
  'customer_order_count' :'int64'}
df_sale_clean = standardize_columns(df_sale_clean)
df_sale_clean = clean_text_col(df_sale_clean, txt_columns)

```

```python
df_sale_clean = convert_datetime(df_sale_clean, columns)
df_sale_clean = convert_numeric(df_sale_clean, rules)
```

```python
df_sale_clean.info()
```

```python
df_sale_clean = remove_duplicate(df_sale_clean)
df_sale_clean = remove_missing(df_sale_clean, name='sale')
df_sale_clean
```

```python
print(valid(df_sale_clean))
```

```python
saved_file(df_sale_clean, '../data/processed/sales_cleaned.csv')
```

# Cleaning Orders_table

```python
df_ord = pd.read_csv("../data/raw/order_items.csv")
df_ord
```

```python
df_ord_clean = standardize_columns(df_ord)
columns = df_ord.select_dtypes(include = ['string']).columns.tolist()

nu_col = {'quantity'           :  'int64', 
   'unit_price'         :  'float64',
 'discount_percentage':  'float64',
   'discount_amount'    :  'float64',
   'gross_sales'        : 'float64',
  'tax_amount'         :  'float64',
  'shipping_cost'      :  'float64',
 'net_sales'          :  'float64',
   'product_cost'       :  'float64',
  'profit'             :  'float64'}
df_ord_clean = clean_text_col(df_ord_clean, columns)
df_ord_clean = convert_numeric(df_ord_clean, nu_col)
df_ord_clean
```

```python
df_ord_clean = remove_duplicate(df_ord_clean)
df_ord_clean = remove_missing(df_ord, 'order')
df_ord_clean
```

```python
print(valid(df_ord_clean))
```

```python
saved_file(df_ord_clean, '../data/processed/orders_clean.csv')
```
